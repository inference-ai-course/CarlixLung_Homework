🧑‍💻 Tesseract Best Practices\
Preprocess Images: Always preprocess images by converting to grayscale, adjusting brightness/contrast, and removing noise to improve Tesseract’s accuracy.

Use Correct --psm: The Page Segmentation Mode (--psm) plays a crucial role in how Tesseract segments the image and interprets the text. Experiment with different modes for complex documents.

Choose the Right Language: Always specify the correct language (-l lang_code) for better accuracy. Tesseract performs poorly when the language is incorrect.

Custom Training: For specialized fonts or languages, you can train Tesseract to recognize custom fonts or languages. This is especially useful for handwriting or unusual fonts.

````python
import arxiv
import trafilatura
import requests
import pytesseract
import json
from pdf2image import convert_from_path
from pathlib import Path
from bs4 import BeautifulSoup

def search_paper(amount=10):
    client = arxiv.Client()
    search = arxiv.Search(
      query = "NLP",
      max_results = amount,
    )
    return [result.entry_id.split('/')[-1] for result in client.results(search)], amount

def html_extraction(arxiv_id):
    try:
        url = f"https://arxiv.org/abs/{arxiv_id}"
        downloaded = trafilatura.fetch_url(url)
        if downloaded:
            soup = BeautifulSoup(downloaded, 'html.parser')
            title = soup.find('h1', class_='title').text.replace('Title:', '').strip()
            abstract = soup.find('blockquote', class_='abstract').text.replace('Abstract:', '').strip()
            authors = [a.text for a in soup.find('div', class_='authors').find_all('a')]
            date = soup.find('div', class_='dateline').text.strip()
            
            print(f"Returning {arxiv_id}")
            return {
                'url': url,
                'title': title,
                'abstract': abstract,
                'authors': authors,
                'date': date
            }
    except:
        pass
    return None

def pdf_extraction(arxiv_id):
    try:
        url = f"https://arxiv.org/pdf/{arxiv_id}.pdf"
        response = requests.get(url, timeout=60)
      
        temp_pdf = "temp.pdf"
        with open(temp_pdf, 'wb') as f:
            f.write(response.content)
        images = convert_from_path(temp_pdf, dpi=300, fmt = 'jpeg')
        text = []
        for image in images:
            page_content = pytesseract.image_to_string(image)
            text.append(page_content)
        output_path = Path('df_ocr') / f'{arxiv_id}.txt'
        output_path.parent.mkdir(exist_ok=True)
        output_path.write_text('\n\n'.join(text), encoding='utf-8')
        
        Path(temp_pdf).unlink()
        print(f"Adding {arxiv_id} to the ocr pdf")
        
        return '\n\n'.join(text)
    except Exception as e:
        print (e)
    return None

def main():
    articles, amount = search_paper(200)
    
    html_papers = []
    for i, article in enumerate(articles, 1):
        arxiv_id = article
        
        html_text = html_extraction(arxiv_id)
        
        if html_text:
            html_papers.append(html_text)
        else:
            print(f"  No HTML, trying OCR...")
        
        pdf_extraction(arxiv_id)
            
    print(f"Out of {amount} papers, there are ",len(html_papers)," papers in HTML.")
    if html_papers:
        with open('arxiv_clean.json', 'w', encoding='utf-8') as f:
            json.dump(html_papers, f, indent=2)
main()
````

````python
import json
from pathlib import Path
import yt_dlp
from faster_whisper import WhisperModel

video_urls = [
    "https://www.youtube.com/watch?v=fLvJ8VdHLA0&pp=ygUbbmF0dXJhbCBsYW5ndWFnZSBwcm9jZXNzaW5n",
    "https://www.youtube.com/watch?v=CMrHM8a3hqw",
    "https://www.youtube.com/watch?v=fNxaJsNG3-s",
    "https://www.youtube.com/watch?v=syvKeNKww_Y&pp=ygUDTkxQ",
    "https://www.youtube.com/watch?v=S3IE9s4tEGQ&pp=ygUDTkxQ",
    "https://www.youtube.com/watch?v=fOvTtapxa9c&pp=ygUDTkxQ",
    "https://www.youtube.com/watch?v=6I-Alfkr5K4&pp=ygUDTkxQ",
    "https://www.youtube.com/watch?v=R-AG4-qZs1A&pp=ygUDTkxQ",
    "https://www.youtube.com/watch?v=Y_hzMnRXjhI",
    "https://www.youtube.com/watch?v=Y_hzMnRXjhI" ## Diplicate to see if data clean successfully


]

#Function to download the audio and title from the youtube vid 
def get_audio_info(url):
    ydl_opts = {
        'format': 'bestaudio/best',
        'postprocessors': [{
            'key': 'FFmpegExtractAudio',
            'preferredcodec': 'mp3',
            'preferredquality': '192',
        }],
        'outtmpl': 'audio_%(id)s.%(ext)s',
        'quiet': False,
    }
    
    with yt_dlp.YoutubeDL(ydl_opts) as ydl:
        info = ydl.extract_info(url, download=True)
        audio_file = f"audio_{info['id']}.mp3"
        title = info.get('title', 'Unknown')
    
    print("Retrived Video Info")

    return {
        'audio_file': audio_file,
        'title': title,
        'video_id': info['id']
    }

#Function to convert audio to string
def audio_to_transcript(audio_info):
    model = WhisperModel("base", device="cpu", compute_type="int8")
    
    print(f"  Transcribing audio...")
    segments, info = model.transcribe(audio_info['audio_file'], beam_size=5)
    
    # Extract segments with timestamps
    segment_list = []
    full_text = []

    for segment in segments:
        segment_list.append({
            'start': segment.start,
            'end': segment.end,
            'text': segment.text.strip()
        })
        full_text.append(segment.text.strip())
    
    transcript_data = {
        'title': audio_info['title'],
        'video_id': audio_info['video_id'],
        'text': ' '.join(full_text),
        'segments': segment_list,
        'language': info.language
    }

    Path(audio_info['audio_file']).unlink(missing_ok=True)

    return transcript_data

def main():
    transcripts = []
    for i, url in enumerate(video_urls,1):
        print("processing video ",i)

        #Function to download the audio and title from the youtube vid 
        audio_info = get_audio_info(url)

        #Function to convert audio to string
        transcript = audio_to_transcript(audio_info)
        transcripts.append(transcript)


    with open('talks_transcripts.jsonl', 'w', encoding='utf-8') as f:
        for i, t in enumerate(transcripts, 1):
            print(f"  Saving video {i}/{len(transcripts)}: {t['title']}")
            f.write(json.dumps(t, ensure_ascii=False) + '\n')
                

main()
````

<h3>Part 4</h3>

````python
from datasketch import MinHash, MinHashLSH
from langdetect import detect
from bs4 import BeautifulSoup
import re
from collections import Counter
import pandas as pd
import json
import os
import csv

#### Prepare the json files for cleaning ####
output_data = []

# 1. Load arXiv JSON data
with open('arxiv_clean.json', 'r', encoding='utf-8') as f:
    arxiv_data = json.load(f)

# 2. Process each arXiv paper
for article in arxiv_data:

    paper_id = article['url'].split('/')[-1]
    txt_filename = f'df_ocr/{paper_id}.txt'
    
    try:
        with open(txt_filename, 'r', encoding='utf-8') as f:
            paper_text = f.read()
        
        output_data.append({
            'id': paper_id,
            'source_type': 'arxiv_paper',
            'title': article['title'],
            'text': paper_text,  
            'authors': ', '.join(article['authors']), 
            'date': article['date'],
            'url': article['url']
        })
    except FileNotFoundError:
        print(f"Warning: Text file not found for {paper_id}")


# 3. Process transcripts
with open('talks_transcripts.jsonl', 'r', encoding='utf-8') as f:
    for line in f:
        talk = json.loads(line)
        
        # Combine all transcript segments into one text
        full_transcript = ' '.join([seg['text'] for seg in talk['segments']])
        
        output_data.append({
            'id': talk['video_id'],
            'source_type': 'youtube_video',
            'title': talk['title'],
            'text': full_transcript,
            'authors': '',  
            'date': '',
            'url': f"https://youtube.com/watch?v={talk['video_id']}"
        })

# 4. Save to CSV
with open('merged_corpus.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.DictWriter(f, fieldnames=['id', 'source_type', 'title', 'text', 'authors', 'date', 'url'])
    writer.writeheader()
    writer.writerows(output_data)

##### Fucntions for data cleaning ####

#Step 1: Filter lang and clean HTML
def clean_html_and_filter_lang(texts, lang='en'):
    filtered = []
    for txt in texts:
        txt = BeautifulSoup(txt, 'html.parser').get_text()
        try:
            if detect(txt.strip()) == lang:
                filtered.append(txt.strip())
        except:
            continue
    return filtered

#Step 2: Minhash dedup
def minhash_deduplication(texts, threshold=0.7):
    lsh = MinHashLSH(threshold=threshold, num_perm=128)
    unique_texts = []
    for i, doc in enumerate(texts):
        m = MinHash(num_perm=128)
        for word in set(doc.split()):
            m.update(word.encode('utf8'))
        if not lsh.query(m):
            lsh.insert(f"doc{i}", m)
            unique_texts.append(doc)
    return unique_texts

#Step 3: Remove PII
def strip_pii(text):
    text = re.sub(r'[\w\.-]+@[\w\.-]+', '[EMAIL]', text)
    text = re.sub(r'\b\d{12,19}\b', '[CREDIT_CARD]', text)
    text = re.sub(r'\b(?:\d{3}-){2}\d{4}\b', '[PHONE]', text)
    return text

#Step 4: Remove n-grams
def remove_repetitive_ngrams(text, n=3, threshold=3):
    words = text.split()
    ngrams = [' '.join(words[i:i+n]) for i in range(len(words)-n+1)]

    counts = Counter(ngrams)
    repetitive = [ngram for ngram, count in counts.items() if count >= threshold]

    for phrase in repetitive:
        # regex-safe version of the phrase
        escaped_phrase = re.escape(phrase)
        # match the phrase repeated 2+ times with optional whitespace
        text = re.sub(rf'(?:{escaped_phrase}\s*){{{threshold},}}', phrase + ' ', text)

    # Remove extra spaces
    text = re.sub(r'\s{2,}', ' ', text).strip()
    return text

try:
    df = pd.DataFrame(output_data)

except:
    print("Reading the saved CSV")
    df = pd.read_csv("merged_corpus.csv")

raw_text = df["text"]

#Some stats for the stat file
print(f"Initial documents: {len(df)}")
raw_texts = df["text"].tolist()
initial_tokens = sum(len(t.split()) for t in raw_texts)
print(f"Initial tokens: {initial_tokens:,}")

print("Deduplicating and cleaning HTML...")
step1 = minhash_deduplication(clean_html_and_filter_lang(raw_text))
print("✅ Done!")

print("Stripping PII...")
step2 = [strip_pii(t) for t in step1]
print("✅ Done!")

print("Removing n-grams....")
cleaned_texts  = [remove_repetitive_ngrams(t) for t in step2]
print("✅ Done!")

#Printing Stat after data cleansing
final_tokens = sum(len(t.split()) for t in cleaned_texts)
print(f"\nFinal documents: {len(cleaned_texts)}")
print(f"Final tokens: {final_tokens:,}")
print(f"Token removal: {((initial_tokens - final_tokens) / initial_tokens * 100):.2f}%")

#Export 
print("Exporting cleaned data...")
with open('clean_corpus.txt', 'w', encoding='utf-8') as f:
    for text in cleaned_texts :
        f.write(text + '\n\n---\n\n')
print("✅ Done!")

#Export Stat
with open('stats.md', 'w') as f:
    f.write("# Data Cleaning Statistics\n\n")
    f.write(f"- Initial documents: {len(raw_texts)}\n")
    f.write(f"- Final documents: {len(cleaned_texts)}\n")
    f.write(f"- Documents removed: {len(raw_texts) - len(cleaned_texts)}\n")
    f.write(f"- Initial tokens: {initial_tokens:,}\n")
    f.write(f"- Final tokens: {final_tokens:,}\n")
    f.write(f"- Token removal: {((initial_tokens - final_tokens) / initial_tokens * 100):.2f}%\n")

````
